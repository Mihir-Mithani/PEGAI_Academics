# 7. Prompting Overview

# 7.1: Experiment : Role-Based Prompting

Formula/Template:

Role + Task + Audience

This prompt assigns a role to the AI and specifies audience context and response structure.

```python
prompt = """
You are a university professor teaching Artificial Intelligence.

Explain Machine Learning to final-year engineering students using:
1. Definition
2. Real-world examples
3. Applications
4. Challenges
"""

response = api_key.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

Observation: The response becomes more professional, educational, and structured.

# 7.2: Experiment – Constraint Prompting

This prompt limits response length and complexity.
Template/Formula: Task + Constraints

```python
prompt = """
Explain Blockchain Technology in less than 120 words using beginner-friendly language.
"""

response = api_key.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)



print(response.text)
```

Observation: The output becomes concise and easier to understand.

# 7.5: Experiment –Hallucination Analysis

AI hallucination refers to the generation of incorrect, misleading, or fabricated information

## Example 1: Hallucination Demonstration

```python
prompt = """
Who won the Nobel Prize in Computer Science in 2023?
Give the winner's name and citation.
"""
```

```python
response = api_key.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

## Example 2: Ambiguous Prompt

```python
prompt = """
Tell me about the founder of Apple.
"""
```

```python
response = api_key.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

The response may discuss only one founder, while Apple had multiple founders:

Steve Jobs
Steve Wozniak
Ronald Wayne

This demonstrates how ambiguity can lead to incomplete or potentially misleading answers.

## Example 3: Fabricated References

```python
prompt = """
Provide 5 research papers published in 2025 about quantum machine learning
with DOI numbers.
"""
```

```python
response = api_key.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

Task:
Verify:



1.   Do the papers exist?
2.   Are the DOI numbers real?
3.   Are the authors correct?

Models sometimes generate realistic-looking but nonexistent citations.

## Example 4: Reducing Hallucination with Better Prompting

```python
prompt = """
Answer only if you are confident.
If you do not know, say 'I don't know'.
Do not invent facts, references, or citations.

Who won the Nobel Prize in Computer Science in 2023?
"""
```

```python
response = api_key.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```

# Class Activity 1

Improve the given bad prompt into good prompt to avoid vague responses from LLM

```python
bad_prompt = """
Give 3 research papers published in 2025 on AI ethics with DOI numbers.
"""

```

```python
good_prompt = """
Give 3 research papers published in 2025 on AI ethics.
Only provide papers if you are certain they exist.
If uncertain, explicitly state that verification is required.
Do not invent citations or DOI numbers.
"""

print("=== Vague Prompt ===")
response1 = api_key.models.generate_content(
    model="gemini-2.5-flash",
    contents=bad_prompt
)
print(response1.text)

```

```python
print("\n=== Improved Prompt ===")
response2 = api_key.models.generate_content(
    model="gemini-2.5-flash",
    contents=good_prompt
)
print(response2.text)
```
