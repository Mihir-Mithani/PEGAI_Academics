# HOMEWORK: Prompt-to-Game Challenge

**Objective**

Create a playable game only through prompts.

**Task 1**

Prompt:

Create a Snake Game in Python.

```python
task1_prompt = """
Create a Snake Game in Python.
"""

response_task1 = api_key.models.generate_content(
    model="gemini-3.6-flash",
    contents=task1_prompt
)

print("=== TASK 1: Basic Prompt Output ===")
print(response_task1.text)
```

**Task 2**
